```python
url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"
df = pd.read_csv(url)
df.head(5)
```

```python
median_value_before = float(df['horsepower'].median())
median_value_before
```

```python
df_after = df.copy()
df_after['horsepower'] = df_after['horsepower'].fillna(median_value_before)
median_value_after = float(df_after['horsepower'].median())
median_value_after
```

```python
df_asia = df.loc[df["origin"] == "Asia"]
df_asia_reduced = df_asia[["vehicle_weight", "model_year"]].head(7)

X = np.array(df_asia_reduced)

XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y)

print(w.sum())
print(w)
```
